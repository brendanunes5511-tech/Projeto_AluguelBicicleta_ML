# Situação-problema: quanto custa alugar uma bicicleta? #

**aluguel de carros**

```python
from sklearn.linear_model import LinearRegression

# ==========================================================
# IDENTIFICAÇÃO
# ==========================================================

nome = "Brenda Nunes da Silva"

print("Nome:", nome)
print("Projeto: Previsão do preço de aluguel de bicicleta")
print("Empresa: Pedal Livre")


# ==========================================================
# DADOS HISTÓRICOS
# ==========================================================

# Duração dos aluguéis em minutos
duracao = [
    [10],
    [15],
    [20],
    [25],
    [30],
    [35],
    [40],
    [50],
    [55],
    [60]
]

# Preços pagos em reais
preco = [
    10.00,
    13.00,
    14.00,
    18.00,
    19.00,
    23.00,
    24.00,
    30.00,
    32.00,
    34.00
]

print("\nDurações:", duracao)
print("Preços:", preco)


# ==========================================================
# CRIAÇÃO DO MODELO
# ==========================================================

modelo = LinearRegression()


# ==========================================================
# TREINAMENTO
# ==========================================================

modelo.fit(duracao, preco)

print("\nModelo treinado com sucesso!")


# ==========================================================
# INFORMAÇÕES DO MODELO
# ==========================================================

print("Coeficiente:", modelo.coef_[0])
print("Intercepto:", modelo.intercept_)


# ==========================================================
# PREVISÃO PARA 45 MINUTOS
# ==========================================================

previsao_45 = modelo.predict([[45]])

print("\nPREVISÃO PRINCIPAL")
print(f"Preço estimado para 45 minutos: R$ {previsao_45[0]:.2f}")


# ==========================================================
# PREVISÃO PARA 20 MINUTOS
# ==========================================================

previsao_20 = modelo.predict([[20]])

print("\nPREVISÃO PARA 20 MINUTOS")
print(f"Preço estimado: R$ {previsao_20[0]:.2f}")
print("Preço histórico: R$ 14.00")


# ==========================================================
# PREVISÃO PARA 60 MINUTOS
# ==========================================================

previsao_60 = modelo.predict([[60]])

print("\nPREVISÃO PARA 60 MINUTOS")
print(f"Preço estimado: R$ {previsao_60[0]:.2f}")
print("Preço histórico: R$ 34.00")


# ==========================================================
# COMPARAÇÃO
# ==========================================================

print("\nCOMPARAÇÃO")
print("=" * 50)

print("20 minutos:")
print("Histórico: R$ 14.00")
print(f"Previsão:  R$ {previsao_20[0]:.2f}")

print()

print("45 minutos:")
print(f"Previsão:  R$ {previsao_45[0]:.2f}")

print()

print("60 minutos:")
print("Histórico: R$ 34.00")
print(f"Previsão:  R$ {previsao_60[0]:.2f}")


# ==========================================================
# PERGUNTAS DE ANÁLISE
# ==========================================================

print("\nPERGUNTAS DE ANÁLISE")
print("=" * 50)


print("\n1. Qual é a variável de entrada e qual é a variável que queremos prever?")
print("A variável de entrada é a duração do aluguel em minutos.")
print("A variável que queremos prever é o preço do aluguel em reais.")


print("\n2. Qual preço o modelo estimou para 45 minutos?")
print(
    f"O modelo estimou aproximadamente R$ {previsao_45[0]:.2f} "
    "para um aluguel de 45 minutos."
)


print("\n3. As previsões para 20 e 60 minutos ficaram próximas dos valores históricos?")

print(
    f"Para 20 minutos, o modelo estimou R$ {previsao_20[0]:.2f}, "
    "enquanto o valor histórico foi R$ 14.00."
)

print(
    f"Para 60 minutos, o modelo estimou R$ {previsao_60[0]:.2f}, "
    "enquanto o valor histórico foi R$ 34.00."
)

print(
    "As previsões ficaram próximas dos valores históricos, "
    "mas não precisam ser exatamente iguais."
)


print("\n4. Por que o preço previsto deve ser apresentado como uma estimativa?")
print(
    "Porque a regressão linear identifica uma tendência geral dos dados."
)

print(
    "Ela não precisa reproduzir exatamente cada preço histórico, "
    "pois existem pequenas variações nos valores."
)


print("\n5. Você teria a mesma confiança em uma previsão para 300 minutos?")
print(
    "Não. Os dados disponíveis vão somente de 10 a 60 minutos."
)

print(
    "Uma previsão para 300 minutos estaria muito distante dos dados "
    "usados no treinamento e, por isso, seria menos confiável."
)


# ==========================================================
# PARECER PARA A GERENTE
# ==========================================================

print("\nPARECER PARA A GERENTE")
print("=" * 50)

print(
    f"Com base nos registros históricos da Pedal Livre e utilizando "
    f"um modelo de regressão linear, o preço estimado para um aluguel "
    f"de 45 minutos é R$ {previsao_45[0]:.2f}. "
    f"O modelo utilizou a duração do aluguel como variável de entrada "
    f"e o preço pago como variável a ser prevista. "
    f"Esse valor deve ser considerado uma estimativa, pois os preços "
    f"históricos apresentam pequenas variações. "
    f"Previsões para períodos muito maiores que os dados disponíveis, "
    f"como 300 minutos, seriam menos confiáveis."
)
```
