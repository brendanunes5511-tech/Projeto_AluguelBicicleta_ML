# Trabalhando com numpy ☻

```python
# como criar um array NumPy? array

array_notas = np.array([7,8,9,10])

# Operações vetorizadas - ADEUS FOR E WHILE!
print(array_notas + 10)
print((array_notas * 10 / 3 + 5) * 2)
```

```python
print(array_notas)
print(array_notas > 8)
# Filtragem -> Mascara de valores
print(array_notas[array_notas > 8])
```

```python
np.sum() #soma todos os numeros do array
np.mean() #media de todos os valores
max() # Valor maximo 
min() # Valor minimo
np.size() # len() - ê a quantidade de elementos.
```

```python
print(max(array_notas))
print(min(array_notas))
print(np.sum(array_notas))
print(np.mean(array_notas))
print(np.size(array_notas))
```
